```python
import json

def load_json_data(filename):
    with open(filename,"r") as file:
        data = json.load(file)
    return data

def user_data(data): 
    print("Users Connections and Updated Data: ")
    for user in data["users"]:
       print(f"ID:{user['id']} {user['name']} having the friendship with the friends {user['friends']}")

    print("")
    for page in data["pages"]:
        print(f"ID: {page['id']} topic: {page['name']}")
```

```python
data_dic = load_json_data("data.json")
```

```python
user_data(data_dic)
```
