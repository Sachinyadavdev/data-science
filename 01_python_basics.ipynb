```python
class Employee:
    company = "HP"
    def __init__(self, name, salary):
       self.name =name
       self.salary = salary

    def print_info(self):
        print(f"The company of {self.name} is {self.company} and the salary of the person is {self.salary}")

    @staticmethod
    def print_name():
        print(f"Company name is {name}")

    @classmethod
    def printClassDetail(cls):
        print(f"The company name is {cls.company}")

emp = Employee("Sachin",48000)
emp.print_info()
emp.print_name()
        
```
