# Financial Data Filtering For Practice


```python
import pandas as pd 

data = pd.read_csv("beginner_pandas_sales.csv")
```

```python
df = pd.DataFrame(data)
```

```python
# print(df)
print(df.head())
# print(df.tail())
# print(df["year"])
# print(df.shape)
# print(df.info)
# print(df.dtypes)
```

## Filtering the Data 

```python
df[["order_id", "customer_name","city","product","quantity","unit_price","order_date"]]

# df.columns
```

```python
df.loc[3]
```

### Selecting Rows and Columns

```python
df.loc[1,"customer_name"]
```

```python
df.iloc[1,2]
```

```python
df.loc[0:2, ["customer_name","city"]]
```

```python
df.at[7,"customer_name"]
```

## Filtering with Conditions


```python
df[df["quantity"]>1]
```

## Multiple Conditions And and OR 

```python
df[(df["quantity"]>1) & (df["city"]=="Pune")]
```

```python
df[(df["quantity"]>2) | (df["city"]=="Pune")]
```

## Quering with .Query

```python
df.query('unit_price > 800 and city=="Pune"') # This will give the query which will be like the Readable Statement.
```

## Dynamic Columns Names 

```python
col = "unit_price"

df.query(f"{col} > 10000")
```

```python
df.query("customer_name == 'Amit'")
```

```python
product_price = 900
df.query("unit_price > @product_price")
```
